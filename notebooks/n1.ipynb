```python
import matplotlib.pyplot as plt
import seaborn as sn

sn.countplot(data=data, x='Churn Label')

plt.title('Customer Churn Distribution')
plt.xlabel('Churn')
plt.ylabel('Number of Customers')

plt.show()
```

```python
sn.histplot(data=data, x='Tenure Months', bins=100)

plt.title('Distribution of Customer Tenure')
plt.xlabel('Tenure (Months)')
plt.ylabel('Number of Customers')

plt.show()
```

Tenure - Customers with shorter tenure churn more.

```python
sn.boxplot(data=data, x='Churn Value', y='Tenure Months')

plt.title('Tenure vs Customer Churn')
plt.xlabel('Churn')
plt.ylabel('Tenure (Months)')

plt.show()
```

```python
data.groupby('Churn Value')['Tenure Months'].median()
```

```python
pd.crosstab(
    data['Contract'],
    data['Churn Value'],
    normalize='index'
) * 100
```
